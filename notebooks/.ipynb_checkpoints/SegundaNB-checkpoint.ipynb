# <span style="color: darkgreen;">Objetos</span>

Hasta ahora hemos estado usando objetos de forma totalmente transparente, casi sin ser conscientes de ello. Pero, en realidad, todo en Python es un objeto, desde números a funciones. El lenguaje provee ciertos mecanismos para no tener que usar explícitamente técnicas de orientación a objetos.

Python permite varios paradigmas de programación, incluyendo la programación orientada a objetos (POO). La POO es una manera de estructurar el código que le hace especiamente efectivo organizando y reutilizando código, aunque su naturaleza abstracta hace que no sea muy intuitivo cuando se empieza.

# <span style="color: darkgreen;">Tuplas</span>

La Tupla es una estructura de datos similar a la lista, pero es **inmutable**, es decir que una vez declarada no puede modificarse.
En lugar de  inicializarse con **corchetes** se hace con **paréntesis**.

### Caracteristicas de las Tuplas

- Las tuplas utilizan menos memoria que las listas.
- Son ideales para usarlas como parámetros en funciones que no queremos sean modificados.
- Pueden ser incorporadas a un diccionario como claves, lo que no es posible hacer con las listas.



```python
tupla_1 = (1, 2, 3) # declaramos una tupla con tres campos
print(tupla_1) # la mostramos a la tupla
print(tupla_1[0])  # accedemos al primer elemento de la tupla
```

### También se pueden declarar sin usar paréntesis

```python
tupla_2 = 2, 4, 3 # declaramos otra tupla ahora sin paréntesis
print(tupla_2) # mostramos el objeto
print(type(tupla_2)) # inspeccionamos la forma de almacenamiento en Python
```

**Veamos ahora ver qué sucede si intentamos modificar la tupla**

```python
tupla_2[0]=12
```

# <span style="color: darkgreen;">Diccionarios</span>

En Python, un diccionario es una estructura de datos que permite almacenar pares clave-valor. Cada elemento en un diccionario está asociado con una clave única que se utiliza para acceder al valor correspondiente. Los diccionarios en Python son mutables, lo que significa que pueden ser modificados después de su creación.

Los diccionarios son útiles cuando necesitas almacenar y acceder a datos de manera eficiente utilizando una clave específica en lugar de un índice numérico, como se haría en una lista o una tupla. Además, los diccionarios son flexibles y pueden contener cualquier tipo de dato como clave y valor.


Un diccionario es una estructura de datos que permite almacenar pares clave-valor. Cada elemento en un diccionario está asociado con una clave única que se utiliza para acceder al valor correspondiente. 

Los diccionarios en Python son mutables, lo que significa que pueden ser modificados después de su creación.

Los diccionarios  se definen mediante llaves {} y los elementos se separan por comas. Cada elemento consiste en una clave seguida de dos puntos: y el valor asociado a esa clave. 

```python
inicial={} # creamos  un diccionario vacío, sin objetos
print(inicial) # lo mostramos
```

```python
# Ahora vamos a incorporar campos al diccionario que llamamos inicial
inicial['color']='rojo'
inicial['pies']=3
inicial['Planeta']='Jupiter'
inicial['edad']=90
inicial
```

```python
inicial['edad']=15 # ahora modificamos un campo del diccionario inicial
inicial # lo mostramos
```

```python
inicial['estatura']='alto'  # ahora agregamos un campo que antes no existía
inicial   # lo mostramos
```

```python
## Creamos una variable para categorizar el campo estatura

## Usamos el if para recorrer las opciones

if inicial['estatura'] =='bajo':
   categoria = 1
elif inicial['estatura'] =='medio':
   categoria = 2
else:
   categoria = 3


print(categoria) # visualizamos el resultado de la inspección
```

```python
# También podemos operar sobre los elementos de un diccionario

inicial['edad']=inicial['edad'] + categoria
inicial
```

```python
inicial['categoria']=categoria # agregamos categoría al diccionario
inicial
```

```python
## Veamos otro ejemplo

# Vamos a crear un diccionario que se llama lenguaje favorito, 
# con cuatro campos donde las llaves son los nombres de los estudiantes
# y los valores son string con el nombre del lenguaje de programación
# preferido por cada uno de ellos

lenguaje_favorito={
    'Luis':'Python',
    'Sara':'R',
    'Eduardo': 'C',
   'Mario': 'Java' 
    }

print(lenguaje_favorito)
```

<span style="color: darkred;">**¿Cómo acceder a cada elemento del diccionario?**<span/>

```python
# accedemos a los items [pares de (llave,valor)] del diccionario

lenguaje_favorito.items() 
```

```python
lenguaje_favorito.keys() # accedemos a las claves
```

```python
lenguaje_favorito.values() # accedemos a  los valores
```

```python
# Inspeccionamos el lenguaje favorito de la segunda estudiante
print("El lenguaje favorito de Sara es " + lenguaje_favorito['Sara'].title()+ '.')  
```

```python
## Vamos a recorrer los items del diccionario lenguaje_favorito
## Vamos a imprimir los elementos para verlos


for Nombre, Lenguaje in lenguaje_favorito.items():
    print("Nombre: " + Nombre.title() + " el lenguaje favorito es "+ Lenguaje.title())
```

```python
## Si queremos visualizar solamente los nombres

for Nombre in lenguaje_favorito.keys():
    print("Nombre: " + Nombre.title())
```

```python
## Y si queremos ordenar los nombres alfabéticamente?? Cómo hacemos??

for Nombre in sorted(lenguaje_favorito.keys()):
    print("Nombre: " + Nombre.title())
```

```python
## Ahora imprimamos los valores en orden alfabético!!

for Lenguaje in sorted(lenguaje_favorito.values()):
    print("Lenguaje: " + Lenguaje.title())
```

```python
for Lenguaje in sorted(lenguaje_favorito.values(), reverse=True):
    print("Lenguaje: " + Lenguaje.title())
```

```python
## Armemos un nuevo diccionario, ahora el valor va a ser una lista
## En la cual ordenamos los lenguajes favoritos de cada estudiante

lenguaje_favorito2={
    'Dante':['Python','C++'],
    'Gabi':['R','Python'],
    'Maia':['C','Java'],
   'Hernan':['Python','C','Pascal'] 
    }

   
lenguaje_favorito2
```

```python
# Declaramos un diccionario para usuario 

usuario={ 'nombre_usuario':'eperez',
    'Nombre':'Enrique',
    'Apellido':'Perez'} 

usuario
```

```python
## Recorremos mediante un ciclo for los elementos de un diccionario

# Atención!!!! Acá en el recorrido hay que considerar la clave y el valor

for k,v in usuario.items():
     # k va a variar sobre la clave y
    # v va a variar sobre el valor
    print("\nClave:" + k)
    print("Valor:" + v)
```

# <font color='darkgreen'> Sets o Conjuntos </font>

Los sets en Python son una estructura de datos usada para almacenar elementos de una manera similar a las listas, pero con algunas diferencias.

Debido a que los conjuntos no pueden tener múltiples ocurrencias del mismo elemento, hace que los conjuntos sean muy útiles para eliminar eficientemente los valores duplicados de una lista o tuple y realizar operaciones de matemáticas comunes como sindicatos e intersecciones.



Para crear un set en Python se puede hacer con <font color='darkred'>Sets o Conjuntos</font>
 y pasando como entrada cualquier tipo iterable, como puede ser una lista. Se puede ver como a pesar de pasar elementos duplicados como dos 8 y en un orden determinado, al imprimir el set no conserva ese orden y los duplicados se han eliminado.

```python
# Crear set Python

s = set([5, 4, 6, 8, 8, 1])
s      
```

```python
print(type(s)) 
```

### Operaciones con sets


A diferencia de las listas, en los set no podemos modificar un elemento a través de su índice. Si lo intentamos, tendremos un TypeError.

```python
s = set([5, 6, 7, 8]) # declaro el set
#s[0] = 3 # intento modificar el primer elemento
```

Los elementos de un set deben ser **inmutables**, por lo que un elemento de un set no puede ser ni un diccionario ni una lista. Si lo intentamos tendremos un TypeError

```python
lista = ["Perú", "Argentina"]
#s = set(["México", "España", lista])
```

**Los sets se pueden iterar de la misma forma que las listas.**



```python
s = set([5, 6, 7, 8])
for i in s:
    print(i) 
    
 
```

   
    
Con la función **len()** podemos saber <font color='darkcyan'>la longitud total del set</font>. Como ya hemos indicado, los duplicados son eliminados.



```python
s = set([1, 2, 2, 3, 4,4,4])
print(len(s)) 

```

También podemos saber si un elemento está presente en un set con el operador in. Si el valor existe en el set, se devolverá True.


```python
s = set(["Guitarra", "Bajo"])
print("Guitarra" in s) 
print("Piano" in s)
```

Los sets tienen además diferentes funcionalidades, que se pueden aplicar
en forma de operador o de método. Por ejemplo, el operador | nos permite realizar la unión de dos sets, lo que equivale a juntarlos. El equivalente es el método union() que vemos a continuación.



```python
s1 = set([1, 2, 3]) # declaramos el primer set
s2 = set([3, 4, 5]) # declaramos el segundo set
print(s1 | s2) # unimos los dos sets y lo mostramos
```

## <font color='darkgreen'> Métodos sets</font>

**El método add() permite añadir un elemento al set.**

s.add(<elem>)

```python
set1 = set([1, 2])
set1.add(3)
print(set1)
```

**El método remove() elimina el elemento que se pasa como parámetro. 
Si no se encuentra, se lanza la excepción KeyError.**

 
s.remove(<elem>)

```python
set2 = set([1, 2, 4, 7])
set2.remove(2)
print(set2) 
```

**El método discard() es muy parecido al remove(), borra el elemento que se pasa como parámetro,y si no se encuentra no hace nada.**

s.discard(<elem>)

```python
set3 = set([1, 2, 5])
set3.discard(3)
print(set3)
```

**El método pop() elimina un elemento aleatorio del set.**

s.pop()

```python
set4 = set([ "A", 2, 9, 0, 1])
set4.pop()
print(set4) 
```

**El método clear() elimina todos los elementos de set.**

s.clear()

```python
set5 = set([1, 8, 6, "B"])
set5.clear()
print(set5)
```

Los sets cuentan con una gran cantidad de métodos que permiten realizar operaciones con dos o más, como la unión o la intersección.

Podemos calcular **la unión entre dos sets usando el método union()**. Esta operación representa la “mezcla” de ambos sets. Nótese que el método puede ser llamado con más parámetros de entrada, y su resultado será la unión de todos los sets.


```python

s1 = {1, 2, 3}
s2 = {3, 4, 5}
print(s1.union(s2)) 
```

También podemos calcular **la intersección entre dos o más set**. Su resultado serán aquellos elementos que pertenecen a ambos sets.

```python
s1 = {1, 2, 3}
s2 = {3, 4, 5}
print(s1.intersection(s2)) 
```

# <font color="darkgreen">Slicing</font> 

La expresión slicing hace referencia a la operación por medio de la cual se extraen elementos de una secuencia, tal como una lista o una cadena de caracteres. Dependiendo del caso, los elementos podrían ser consecutivos o podrían estar separados dentro de la secuencia original. Dado que esta operación es de uso bastante frecuente, Python ofrece características que hacen posible construir slices de forma breve, sin tener que construir funciones especializadas para hacerlo.

Supongamos que tenemos la siguiente cadena de caracteres con la que inicia el himno de Argentina:

```python
cadena = '¡Oid mortales el grito sagrado: Libertad,Libertad,Libertad!'
cadena
```

Cómo podemos hacer para extraer elementos de esta cadena usando paréntesis cuadrados y una posición numerada desde cero.

```python
cadena[0] # estamos extrayendo un slice de la cadena original con el primer caracter
```

```python
cadena[1:4] # extraemos los caracteres desde la posición 1 hasta la 4 (excluida), 
```

```python
cadena[30:] # recortamos de la posición 30 en adelante
```

```python
cadena[:30] # recortamos hasta la posición 30 exclusive
```

```python
cadena[None:30] # otra forma de hacer lo mismo que la anterior
```

```python
# Crear una copia de la cadena original
cadena2 = cadena[:]
cadena2
```

**Slicing intermedio: posiciones negativas**

Algo muy interesante en Python con respecto a los slices, es que las posiciones de inicio y de fin no necesariamente tienen que ser números positivos: si se usan números negativos, Python asume que se están referenciando las posiciones numeradas desde la última hasta la primera, empezando con -1

```python
cadena3= 'HOLA, GENTE!'
cadena3[-1] # consultamos la última posición

```

```python
cadena3[-2] # consultamos la anteúltima posición
```

```python
cadena3[-10: -6] #¿qué hace esto?
```

```python
cadena3[::-1] #¿como recorre esta instrucción la cadena?
```

 <span style="color: darkgreen;">**Hay operaciones que matemáticamente no existen...sin embargo podrían interesarnos en la práctica.** <span/>

SQRT significa square root en inglés.

```python
M=(abs(N)) # calcula el valor absoluto de los elementos de la matriz N y
# los guarda en la matriz M

display(Math(sym.latex(sym.sympify(M)))) # muestra la cara bonita de M
```

```python
Q=np.sqrt(M) # calcula la raiz cuadrada de cada elemento de la matriz M 
# y los guarda en la matriz Q
display(Math(sym.latex(sym.sympify(Q))))# muestra la cara bonita de Q
```

```python
R=Q**2 # calcula el cuadrado de cada elemento de la matriz Q
# y los guarda en la matriz R




```

Acá hay que hacer una salvedad, esto no es el cuadrado de la matriz Q.

```python

display('R',Math(sym.latex(sym.sympify(R)))) # muestra la cara bonita de R
```

# <font color="darkgreen">DataFrames</font> 

A diferencia de las Series, que son objetos correspondientes a paneles unidimensionales, los DataFrames son paneles bidimensionales compuestos por filas y columnas, que permiten destacar las relaciones entre las distintas variables de la serie de datos. Un DataFrame es una serie de Series Pandas indexadas por un valor.

# <font color="darkcyan">¿Cómo se presenta un DataFrame?</font>

El formato de estas estructuras puede compararse con los diccionarios de Python. Efectivamente, las claves son los nombres de las columnas y los valores son las Series. Su estructura puede considerarse similar a la de una hoja de cálculo de Excel.

Cada fila contiene datos específicos de varias columnas, que son variables. El nombre de las filas de un DataFrame se llama «index» que, por defecto, empieza siempre por 0.

Para poder trabajar con un DataFrame importamos la librería pandas. 
Pandas es una de las principales librerías de Python. Las estructuras más utilizadas en este módulo son  los DataFrames.

```python
## Armamos un dataframe a partir de un diccionario


##importamos el paquete pandas
#!pip install pandas 

import pandas as pd

datos= {'nombre':["Luis","María","Pepe"],
        'ciudad':["San José","Lima", "Buenos Aires"],
        'edad':[23,56,12]
        }                          # creamos un diccionario de datos
datos_pandas= pd.DataFrame(datos)  # lo convertimos en dataframe
print(datos_pandas)                # lo visualizamos
```

```python
datos_pandas.info()   # nos da la información de qué almacenamos en el df
```

```python
datos_pandas.size  ## cantidad de datos o entradas.
```

```python
datos_pandas.shape  ## Dimensión. Como el dim de R
```

```python
datos_pandas.shape[0]  ## número de filas
```

```python
datos_pandas.shape[1] ## número de columnas
```

```python
#  Armamos un dataframe a partir de una lista de edades


datos2= [['Alejandro',10],['Roberto',12],['Claudio',13]] #ingresamos los datos
DF= pd.DataFrame(datos2,columns=['Nombre','Edad']) # ponemos nombre a las columnas
print(DF) # visualizamos el dataframe
```

```python
# Construimos ahora el dataframe como combinación de variables

# Un dataframe se construye  .. y ya quedan los nombres de columnas
datos3={'Nombre':['Tomás','Javier','Esteban','Ricardo'],
        'Edad':[28,34,29,42]
        }
DF= pd.DataFrame(datos3, index=['ind1','ind2','ind3','ind4']) # index es equivalente a un row name en R
print(DF)
```

```python
print(DF[1:3]) #visualizamos las filas 2 y 3 del dataframe
```

```python
print(DF.loc['ind2']) # accedemos a una fila en particular del dataframe
```

```python
print(DF.iloc[1])  ### en forma equivalente por posición
```

```python
print(DF.loc[:,'Nombre']) # accedemos a una columna del dataframe
```

```python
# con los ":" quiere decir todas la filas de la columna Nombre
# es una forma de extraer más específicamente
# Es muy potente
print(DF.loc['ind2':'ind3','Nombre'])
```

```python
# Cambiar la edad de Tomás a 15
DF.loc[DF['Nombre'] == 'Tomás', 'Edad'] = 15
print(DF)
```

### Agregamos leyendas para distinguir las gráficas

```python
fig, ax = plt.subplots()

ax.plot(x, sin, label='sin')

ax.plot(x, cos, label='cos')

ax.legend()
```

```python
### Podemos incluir símbolos matemáticos en las leyendas
```

```python
fig, ax = plt.subplots()
ax.plot(x, sin, label='$f_1(x) = sin(x)$')

ax.plot(x, cos, label='$f_2(x) = cos(x)$')


ax.legend()
```

### Ubicación de la leyenda

```python

fig, ax = plt.subplots()
ax.plot(x, sin, label='$f_1(x) = sin(x)$')

ax.plot(x, cos, label='$f_2(x) = cos(x)$')


ax.legend()
ax.legend(loc='upper right')
```

### Aplicando estilos al gráfico

```python
fig, ax = plt.subplots()
ax.plot(x, sin, label='$f_1(x) = sin(x)$')

ax.plot(x, cos, label='$f_2(x) = cos(x)$')

sin_style = dict(linewidth=3, color='darkorange')
cos_style = dict(marker='o', markerfacecolor='limegreen', color='darkgreen')

ax.plot(x, sin, label='$f_1(x) = sin(x)$', **sin_style)

ax.plot(x, cos, label='$f_2(x) = cos(x)$', **cos_style)
```

```python
## Acotando ejes

fig, ax = plt.subplots()
ax.plot(x, sin, label='$f_1(x) = sin(x)$')

ax.plot(x, cos, label='$f_2(x) = cos(x)$')

sin_style = dict(linewidth=3, color='darkorange')
cos_style = dict(marker='o', markerfacecolor='limegreen', color='darkgreen')

ax.plot(x, sin, label='$f_1(x) = sin(x)$', **sin_style)

ax.plot(x, cos, label='$f_2(x) = cos(x)$', **cos_style)
 
ax.set_xlim(0, np.pi) #límites del eje x
ax.set_ylim(-1, 1) # límites del eje y

ax.grid()  # sólo a efectos estéticos
```
