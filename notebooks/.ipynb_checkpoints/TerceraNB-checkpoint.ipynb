# <font color="darkgreen">Estadística Descriptiva </font>

La estadística descriptiva: La cual se dedica a recolectar, ordenar, analizar y representar a un conjunto de datos, con el fin de describir apropiadamente las características de este. Calcula los parámetros estadísticos que describen el conjunto estudiado. Algunas de las herramientas que utiliza son gráficos, medidas de frecuencias, medidas de centralización, medidas de posición, medidas de dispersión, entre otras.

# <font color="#229955">Test para Comparar Medias</font>

**Muestras Independientes**

Hipótesis contrastadas


Las hipótesis contrastadas por el t-test de muestras independientes son:

$H_0$
 : no hay diferencias entre las medias:  $\mu_x=\mu_y$
 
$H_1$
 : sí hay diferencias entre las medias:  $\mu_x \neq \mu_y$
 
 El estadístico utilizado por el t-test de muestras independientes se calcula como:

$t=\dfrac{\overline{x}-\overline{y}}{\sqrt{\dfrac{s^2_x}{n_x}+\dfrac{s^2_y}{n_y}}} \sim t_{n_x+n_y-2}$

donde $\overline{x}$ y $\overline{y}$ son las medias muestrales,

$s^2_x$ y $s^2_y$ son las varianzas muestrales y 

$n_x$ y $n_y$ son los tamaños muestrales.

Este estadístico sigue una distribución conocida como distribución t de Student similar a la distribucion normal pero con mayor peso en las colas.

Esta distribución tiene un parámetro vinculado a los tamaños muestrales que se denomina grados de libertad.

En este caso los grados de libertad son $n_x+n_y-2$.

Cuanto mayor es el número de grados de libertad más próxima a la normal se hace la distribución t de Student.

Esto vale si pueden suponerse que las varianzas de las dos poblaciones que se comparan son iguales, en caso contrario el cálculo de grados de libertad es un poco más complejo (ver test de Welch).

Recordemos los pasos para realizar  un t-test con el objetivo de determinar si la diferencia entre medias es significativa son:

 a) Formular las hipótesis (nula y alternativa).

 b) Calcular el estadístico de contraste.

 c) Seleccionar el tipo de test (unilateral o bilateral).

 d) Elegir el nivel de significancia  $\alpha$.
 
 e) Calcular el p valor y compararlo con el nivel de significación establecido.

 f) Decidir y concluir en el contexto del problema planteado.


### <font color="brown">Primer Ejemplo</font>

```python
# Importamos las librerías

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python

# Vamos a cargar unos datos muy conocidos acerca de lirios 
iris = sns.load_dataset('iris')

# Guardamos en dos bases distintas las especies setosa y versicolor
setosa = iris[iris['species'] == 'setosa']
versicolor = iris[iris['species'] == 'versicolor']

# guardamos en dos variables las longitudes de petalos de estas dos especies
setosa_petal_lengths = setosa['petal_length']
versicolor_petal_lengths = versicolor['petal_length']
```

Verificamos la normalidad de las variables en los grupos elegidos

```python
# Shapiro-Wilk test
# ==============================================================================
shapiro_test1 = stats.shapiro(setosa_petal_lengths )
print("estadistico=",round(shapiro_test1.statistic,3),
"p_valor=",round(shapiro_test1.pvalue,3))



```

```python
# Shapiro-Wilk test
# ==============================================================================
shapiro_test2 = stats.shapiro(versicolor_petal_lengths )
print("estadistico=",round(shapiro_test2.statistic,3),
"p_valor=",round(shapiro_test2.pvalue,3))
```

Como no se rechaza la normalidad podemos aplicar la prueba t

```python


# Aplicamos la prueba t
t_stat, p_value = stats.ttest_ind(setosa_petal_lengths, versicolor_petal_lengths)

```

```python
# ahora interpretamos los resultados 
# Fijamos el nivel de significación
alpha = 0.05
if p_value < alpha:
    print("Rechazamos la hip.nula, las diferencias de media son significativas.")
else:
    print("no rechazamos la hip. nula; las medias no difieren significativamente.")
```

## Si los datos no son normales 

O en caso que  $n_1$ o $n_2$ sean menores que 30, debemos aplicar la prueba de Wilcoxon Mann-Whitney.

En este ejemplo vamos a suponer que tenemos datos diagnósticos de cuatro mujeres y cinco hombres. Todos fueron diagnosticados con diabetes y tenemos la edad a la cual se les descubrió la enfermedad. Queremos saber si hay diferencia en la edad entre hombres y mujeres. Los datos son:

```python
from scipy.stats import mannwhitneyu

Hombres = [19, 22, 16, 29, 24]
Mujeres = [16, 11, 17, 12]


# Realizar la prueba de Wilcoxon-Mann-Whitney
statistic, p_value = mannwhitneyu(Hombres, Mujeres)

# Imprimir el resultado
print("Estadístico de prueba:", statistic)
print("Valor p:", p_value.round(3))
```

# <font color="#229955">Test para Comparar Proporciones</font>

## <font color="blue">Hipótesis a evaluar</font>

**Hipótesis nula**

$H_0: p_1=p_2$

No existen diferencias entre las proporciones de las poblaciones ($p_1– p_2=0$).

La proporción de sujetos  que satisfacen la condición es la misma en  las dos poblaciones

**Hipótesis alternativa** 

Podemos seleccionar una de las siguientes hipótesis alternativas:

$H_1: p_1 ≠  p_2$

La diferencia entre las proporciones de población ($p_1– p_2$) no es igual a la diferencia hipotética ($d_0$).

De los sujetos que satisfacen la condición, existen diferencias en la proporción de la poblacion 1 ($p_1$) y la poblacion 2 ($p_2$).

$H_1: p_1 > p_2$

La proporción de la población 1 (p1) es mayor que la proporción de la población 2 ( $p_2$).

De los sujetos que satisfacen la condición, a proporción de la poblacion 1  ($p_1$) es mayor que la proporción de la población 2 ($p_2$).

$H_1: p_1 < p2$

La proporción de la población 1 ($p_1$) es menor que la proporción de la población 2 ($p_2$).

De los sujetos que satisfacen la condición, la  proporción de la poblacion 1 ($p_1$) es menor que a proporción de la poblacion 2 ($p_2$).

NOTA: Si elegimos la primer opción estaremos ante una prueba bilateral, mientras que la segunda y tercera son pruebas unilaterales (derecha e izquierda, respectivamente).

```python
from statsmodels.stats.proportion import proportions_ztest

# Datos de ejemplo
exito_gr1 = 40  # Número de éxitos en el grupo 1
n_1 = 100    # Tamaño de la muestra del grupo 1
exito_gr2 = 30  # Número de éxitos en el grupo 2
n_2 = 100    # Tamaño de la muestra del grupo 2

# Realizar la prueba de proporciones z
stat, pval = proportions_ztest([exito_gr1, exito_gr2], [n_1, n_2])

# Imprimir el resultado
print("Estadístico de prueba:", stat.round(3))
print("Valor p:", pval.round(3))
```

# **Ejercicio 22:**

En un estudio se quiere comparar la proporción de hombres y mujeres que completan un curso en línea. Tienes dos grupos de participantes: uno de hombres y otro de mujeres. Se quiere determinar si hay una diferencia significativa en la proporción de hombres y mujeres que completan el curso.


    En el grupo de hombres, 70 de 200 completaron el curso.
    En el grupo de mujeres, 35 de 180 completaron el curso.
    
 Decidir y concluir con una significacion del 5%. 

```python
from statsmodels.stats.proportion import proportions_ztest

# Datos de ejemplo
exito_gr1 = 70  # Número de éxitos en el grupo 1
n_1 = 200     # Tamaño de la muestra del grupo 1
exito_gr2 = 55  # Número de éxitos en el grupo 2
n_2 = 180     # Tamaño de la muestra del grupo 2

# Realizar la prueba de proporciones z
stat, pval = proportions_ztest([exito_gr1, exito_gr2], [n_1, n_2])

# Imprimir el resultado
print("Estadístico de prueba:", stat.round(3))
print("Valor p:", pval.round(3))
```

```python
# Para rechazar H0 el p valor debe ser menor a 0.05. En este caso no lo es, por lo tanto no hay evidencia para rechazar que p1=p2
```
