### Estructuras de Repetición : for

```python
# Ejemplo for en lista
# Permite recorrer todos los elementos de una lista
# En el mismo for se le pone nombre a la variable que almacena cada elemento
# En este caso la lista se llama mis_numeros y cada elemento es un numero

# El objetivo del código es sumar todos los elementos de la lista
# El acumulador debe inicializarse en 0
# De lo contrario guardará el valor en caso de hacerse otras corridas

suma=0
mis_numeros=[4,8,12,-18] 
#la variable numero designará a los elementos de la lista
for numero in mis_numeros:
    suma=suma+numero
    print("la suma parcial es ",numero)
    


print()
print("esto se imprime fuera del ciclo")
print('La lista es',mis_numeros)
print("La suma de los números de la lista es:")
print(suma)
```

# Otro operador cíclico while

La estructura while permite ejecutar un bloque de código repetidamente mientras se cumpla una condición específica. Esto proporciona una flexibilidad considerable ya que el número de iteraciones no está predefinido y puede depender de los datos de entrada. En algunas situaciones, la estructura while puede ser más eficiente que la estructura for, especialmente cuando no se conoce de antemano el número exacto de iteraciones necesarias o cuando se requiere una condición de salida del ciclo que no puede ser fácilmente expresada en un rango fijo.

Su sintaxis es

while <span style="color: darkred;">condición </span> : 

____<span style="color: darkgreen;">cuerpo del while </span>
    #indentado con 4 espacios 

```python
# Ejemplo de ciclo while con un contador
 
# El objetivo del código es imprimir los números del 7 al 10
contador = 7            # inicializa el contador
while contador <= 10:
    print(contador)
    contador += 1      # es equivalente a colocar contador=contador + 1
```

## Otro ejemplo con while

* Ingrese un número y calcule e imprima el factorial.
* Nota: Recordar que así como las sumatorias se inicializan en 0 las productorias se inicializan en 1.

```python

nro=int(input('Ingrese un número natural: '))
produc=1                     # inicializa el contador
contador=1                   # inicializa el contador
while contador < nro:
    contador += 1 
    produc=produc * contador
print("Factorial= ", produc)
```

# **Ejercicio 9:**

Escribe un programa que solicite al usuario un número entero positivo y luego imprima un contador regresivo desde ese número hasta 1

```python
# Ejemplo de ciclo while con un contador
 
# El objetivo del código es imprimir los números del 10 al 1
contador = 10            # inicializa el contador
while contador >= 1:
    print(contador)
    contador -= 1      # es equivalente a colocar contador=contador - 1
```
